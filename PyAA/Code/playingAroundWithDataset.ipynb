```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors
import seaborn as sns
import matplotlib.ticker as ticker
plt.rcParams['figure.dpi'] = 120
data = pd.read_csv('../data/PSCompPars_2023.03.06_12.57.46.csv', comment='#', low_memory=False)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
#data
```

```python
test = ['Transit', 'Radial Velocity', 'Imaging', 'Microlensing']
datatest = data.loc[data['discoverymethod'].isin(test)]
```

```python
graph = sns.scatterplot(data = datatest, x = 'pl_orbsmax', y = 'pl_bmasse', hue = 'discoverymethod')
graph.set(xscale = 'log', yscale = 'log', xlabel = 'Semimajor Axis [AU]', ylabel = 'Mass [Earth Mass]')
graph.set_xlim(0, 100)
graph.set_ylim(0, 10000)
```

```python
giant = data.loc[data['pl_radj'] > 1]
hj = giant.loc[giant['pl_eqt'] > 999]
```

```python
import matplotlib.pyplot as plt
#plt.scatter(hj['st_teff'], hj['pl_orbincl'])
y_error = [hj['pl_orbinclerr1'], -hj['pl_orbinclerr2']]
plt.errorbar(x = hj['st_teff'], y = hj['pl_orbincl'], yerr = y_error,fmt='x',elinewidth = 1, color="r")
plt.title("Hot Jupiters Stellar Temp over Inclination")
plt.xlabel("Inclination [deg]")
plt.ylabel("Stellar Effective Temperature [K]")
plt.show()
```

```python
byHost = data.groupby(by='hostname')
count = byHost.count().sort_values(by='pl_name')
plancount = sns.countplot(data = count, x = 'pl_name', hue= 'st_spectype')
plancount.set(yscale='log')
plancount.set_xlabel('number of planets per system')
plancount.set_ylabel('number of systems')
```

```python
boxplot = sns.boxplot(data = data, y = 'discoverymethod', x = 'pl_orbper')
boxplot.set_xlabel("Orbital Period [Earth Days]")
boxplot.set_xlim(1, 1000000)
boxplot.set_xscale('log')
whatIsThis = data[data['discoverymethod'].str.contains("Disk Kinematics")]
whatIsThis
```

```python
test = sns.displot(data,x = 'pl_eqt', kde = True, bins = 5)
test.set_xlabels('Temperature [K]')
test.set(yscale= 'log')
```

```python
tempplanets = data.loc[data['pl_eqt'] > 1]

tempplanets['pl_eqt_bin'] = pd.cut(tempplanets['pl_eqt'], bins = [1, 800, 1600, 2400, 3200, 4000])
tempplanets['Avg'] = tempplanets.groupby('pl_eqt_bin')['pl_eqt'].transform("mean")
averaged = sns.countplot(data = tempplanets, x = 'Avg', hue = 'pl_eqt_bin')
averaged.set(xlabel='average temp per group [K]', ylabel= 'count',yscale='log')
averaged.set_xticklabels([551.8, 1097.1, 1947.5, 2605.3, 3499.5])
averaged.legend(title = 'temperature range [K]')
```

```python
dist = sns.scatterplot(data = data, x = 'pl_orbsmax', y = 'pl_eqt', hue = 'st_spectype')
dist.set(xlabel = 'Semimajor axis [AU]', ylabel = 'planet temperature [K]')
dist.set_xscale('log')
dist.set_yscale('log')
dist.legend_.remove()
```
